```python
import os

# Directory principale dove si trovano i file Kotlin
directory = r'app'
output_file = 'merged_java_app.java'

# Funzione per ottenere solo la directory principale specificata dall'utente
def get_base_directory_name(base_directory):
    return os.path.basename(os.path.normpath(base_directory))

# Apri il file di output in modalità scrittura
with open(output_file, 'w') as outfile:
    base_directory_name = get_base_directory_name(directory)
    for foldername, subfolders, filenames in os.walk(directory):
        for filename in filenames:
            if filename.endswith('.java'):
                filepath = os.path.join(foldername, filename)
                # Scrivi il commento con la cartella principale
                outfile.write(f"\n// Directory: {base_directory_name}\n")
                # Scrivi anche il percorso della sottocartella da cui viene preso il file
                relative_path = os.path.relpath(foldername, directory)
                outfile.write(f"// Subdirectory: {relative_path}\n")
                outfile.write(f"// File: {filename}\n\n")
                # Aggiungi il contenuto del file DART
                with open(filepath, 'r') as infile:
                    outfile.write(infile.read())
                    outfile.write("\n")

print(f'I file java sono stati uniti in {output_file}')
```
