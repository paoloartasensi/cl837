```python
import os
import subprocess

# Directory di partenza per la ricerca
start_directory = 'lib'  # Sostituisci con il percorso della tua directory Dart
#start_directory = r'C:\Users\UserDemo\Documents\VScode\WATCHFIT_app\lib\features\workout'

# File unico in cui salvare tutti i contenuti .dart
output_file = 'merged_files_APP.txt'

# Percorso del file pubspec.yaml
pubspec_file = 'pubspec.yaml'

# Funzione per ottenere il nome del branch Git attualmente attivo
def get_current_git_branch():
    try:
        branch_name = subprocess.check_output(['git', 'rev-parse', '--abbrev-ref', 'HEAD'],
                                             stderr=subprocess.DEVNULL).strip().decode('utf-8')
        return branch_name
    except Exception as e:
        print(f"Errore nell'ottenere il branch Git: {e}")
        return "unknown_branch"

# Funzione per cercare e copiare il contenuto dei file .dart e includere il pubspec.yaml
def merge_dart_files_with_pubspec(start_directory, output_file, pubspec_file):
    branch_name = get_current_git_branch()

    with open(output_file, 'w') as outfile:
        # Scrivi l'intestazione con il nome del branch Git
        outfile.write(f'// Merged files from branch: {branch_name}\n\n')

        # Aggiungi il contenuto di pubspec.yaml se esiste
        if os.path.exists(pubspec_file):
            with open(pubspec_file, 'r') as pubspec:
                outfile.write('// --- Start of pubspec.yaml ---\n')
                outfile.write(pubspec.read())
                outfile.write('\n// --- End of pubspec.yaml ---\n\n')
        else:
            print(f"File {pubspec_file} non trovato. Procedo senza di esso.")

        # Aggiungi i file .dart
        for root, dirs, files in os.walk(start_directory):
            for file in files:
                if file.endswith('.dart'):
                    file_path = os.path.join(root, file)
                    with open(file_path, 'r') as infile:
                        # Scrivi un'intestazione per ogni file nel file di output
                        outfile.write(f'// --- Start of {file_path} ---\n')
                        outfile.write(infile.read())
                        outfile.write(f'\n// --- End of {file_path} ---\n\n')
    print(f'Tutti i file .dart e il pubspec.yaml (se presente) sono stati copiati in {output_file}')

# Esegui la funzione
merge_dart_files_with_pubspec(start_directory, output_file, pubspec_file)
```

## PDF CREATOR

```python
import os
import subprocess
from fpdf import FPDF

# Directory di partenza per la ricerca
start_directory = 'lib'

# File PDF di output
output_pdf_file = 'merged_files_APP.pdf'

# Percorso del file pubspec.yaml
pubspec_file = 'pubspec.yaml'


def get_current_git_branch():
    """
    Ritorna il nome del branch Git attualmente attivo.
    """
    try:
        branch_name = subprocess.check_output(
            ['git', 'rev-parse', '--abbrev-ref', 'HEAD'],
            stderr=subprocess.DEVNULL
        ).strip().decode('utf-8')
        return branch_name
    except Exception as e:
        print(f"Errore nell'ottenere il branch Git: {e}")
        return "unknown_branch"


def merge_dart_files_with_pubspec_to_pdf(start_directory, output_pdf_file, pubspec_file):
    """
    Cerca tutti i file .dart in 'start_directory' e unisce i loro contenuti,
    insieme a quello di pubspec.yaml, in un PDF (A4) con PyFPDF.
    """

    branch_name = get_current_git_branch()

    # Qui accumuleremo tutto il testo prima di generare il PDF
    text_content = ""

    # 1. Intestazione con il nome del branch
    text_content += f"// Merged files from branch: {branch_name}\n\n"

    # 2. Aggiungi il contenuto del pubspec.yaml se esiste
    if os.path.exists(pubspec_file):
        text_content += "// --- Start of pubspec.yaml ---\n"
        with open(pubspec_file, 'r', encoding='utf-8') as pubspec:
            text_content += pubspec.read()
        text_content += "\n// --- End of pubspec.yaml ---\n\n"
    else:
        print(f"File {pubspec_file} non trovato. Procedo senza di esso.")

    # 3. Aggiungi i file .dart
    for root, dirs, files in os.walk(start_directory):
        for file in files:
            if file.endswith('.dart'):
                file_path = os.path.join(root, file)
                with open(file_path, 'r', encoding='utf-8') as infile:
                    text_content += f"// --- Start of {file_path} ---\n"
                    text_content += infile.read()
                    text_content += f"\n// --- End of {file_path} ---\n\n"

    # A questo punto abbiamo tutto il testo in 'text_content'.
    # Creiamo il PDF con PyFPDF.
    pdf = FPDF(format='A4')    # formato A4
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()
    pdf.set_font("Courier", size=10)  # Usa Courier (monospazio) o "Arial", come preferisci.

    # Spezzettiamo il testo in righe e stampiamole nel PDF
    for line in text_content.split('\n'):
        pdf.multi_cell(0, 5, line)  # 0 = larghezza cella automatica, 5 = altezza
    # Salviamo il PDF
    pdf.output(output_pdf_file)
    print(f"PDF generato con successo: {output_pdf_file}")


# Esegui la funzione
merge_dart_files_with_pubspec_to_pdf(start_directory, output_pdf_file, pubspec_file)
```
